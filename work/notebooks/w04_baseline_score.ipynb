```python
import os
import pandas as pd
import numpy as np
from huggingface_hub import snapshot_download
from google.colab import userdata
```

```python
march_path = os.path.join(
    repo_path,
    "fact_content_daily_performance",
    "month=2026-03",
    "data_0.parquet"
)

df = pd.read_parquet(march_path)

print("Rows:", len(df))
print("Columns:", len(df.columns))
print("Date range:", df["report_date"].min(), "to", df["report_date"].max())

df.head()
```

```python
content_path = os.path.join(
    repo_path,
    "dim_content.parquet"
)

dim_content = pd.read_parquet(content_path)

print("Content dimension shape:", dim_content.shape)
print("\nColumns:")
print(dim_content.columns.tolist())

dim_content.head()
```

```python
# Check how many rows exist per content page
content_counts = (
    dim_content
    .groupby("content_hash_id")
    .size()
    .describe()
)

print(content_counts)
```

```python
# Check the date ranges available in the content dimension
print("Created date range:")
print(dim_content["content_created_date"].min(),
      "to",
      dim_content["content_created_date"].max())

print("\nUpdated date range:")
print(dim_content["content_updated_date"].min(),
      "to",
      dim_content["content_updated_date"].max())
```

```python
# Check whether the content dimension has duplicate content IDs
duplicate_content = dim_content["content_hash_id"].duplicated().sum()

print("Duplicate content_hash_id rows:", duplicate_content)
```

```python
# Aggregate March performance to one row per content page
march_pages = (
    df.groupby(["client_hash_id", "content_hash_id"], as_index=False)
      .agg(
          impressions_30d=("gsc_impressions", "sum"),
          clicks_30d=("gsc_clicks", "sum"),
          avg_position=("gsc_avg_position", "mean"),
          days_with_gsc=("gsc_data_available", "sum")
      )
)

print("Pages:", len(march_pages))
march_pages.head()
```

```python
content_cols = [
    "client_hash_id",
    "content_hash_id",
    "content_created_date",
    "content_updated_date",
    "search_volume",
    "word_count",
    "is_published",
    "is_deleted"
]

march_pages = march_pages.merge(
    dim_content[content_cols],
    on=["client_hash_id", "content_hash_id"],
    how="left"
)

print("Rows after merge:", len(march_pages))
march_pages.head()
```

```python
decision_date = pd.Timestamp("2026-03-31")

march_pages["content_updated_date"] = pd.to_datetime(
    march_pages["content_updated_date"]
)

# Only updates known by the March 31 decision date
march_pages["known_update_date"] = march_pages["content_updated_date"].where(
    march_pages["content_updated_date"] <= decision_date
)

march_pages["days_since_update"] = (
    decision_date - march_pages["known_update_date"]
).dt.days

print("Rows with a known update date:", march_pages["days_since_update"].notna().sum())
print("Rows without a known update date:", march_pages["days_since_update"].isna().sum())

print("\nDays since update:")
print(march_pages["days_since_update"].describe())
```

```python
march_pages["staleness_bucket"] = pd.cut(
    march_pages["days_since_update"],
    bins=[-1, 90, 180, 365, float("inf")],
    labels=[
        "0-90 days",
        "91-180 days",
        "181-365 days",
        "365+ days"
    ]
)

staleness_table = (
    march_pages["staleness_bucket"]
    .value_counts()
    .sort_index()
    .rename_axis("staleness_bucket")
    .reset_index(name="n")
)

staleness_table
```

```python
staleness_summary = (
    march_pages
    .groupby("staleness_bucket", observed=False)
    .agg(
        n=("content_hash_id", "size"),
        median_impressions=("impressions_30d", "median"),
        median_clicks=("clicks_30d", "median")
    )
    .reset_index()
)

staleness_summary
```

## 1. My rule and its reason codes

Baseline rule: Prioritize published content that has been updated a long time ago and still has meaningful search visibility. The score combines content staleness and recent impressions so that older pages with stronger observed search visibility receive higher priority for a refresh review.

Reason codes:

STALE_VISIBLE — content is older and has meaningful search impressions.
STALE_LOW_VISIBILITY — content is older but has low search visibility.
RECENT_VISIBLE — content is relatively recent but has meaningful search visibility.
LOW_PRIORITY — does not meet the stronger staleness or visibility conditions.

Action: refresh_review for the highest-priority pages; lower-scoring pages remain monitor.

This is deliberately a baseline rule, not an ML model.

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
# Build the baseline ranked queue

queue = march_pages.copy()

# Fill missing values used by the rule
queue["days_since_update"] = queue["days_since_update"].clip(lower=0)
queue["impressions_30d"] = queue["impressions_30d"].fillna(0)

# Define the two baseline signals
# Staleness: older content gets a higher value
# Visibility: more recent impressions indicate stronger observed search visibility
queue["staleness_score"] = queue["days_since_update"]

queue["visibility_score"] = (
    queue["impressions_30d"].clip(upper=10000) / 100
)

# Combined baseline score
queue["score"] = (
    0.5 * queue["staleness_score"]
    + 0.5 * queue["visibility_score"]
)

# Define simple baseline conditions
stale = queue["days_since_update"] > 90
visible = queue["impressions_30d"] >= 100

# Assign exactly one reason code
queue["reason_code"] = "LOW_PRIORITY"

queue.loc[stale & visible, "reason_code"] = "STALE_VISIBLE"
queue.loc[stale & ~visible, "reason_code"] = "STALE_LOW_VISIBILITY"
queue.loc[~stale & visible, "reason_code"] = "RECENT_VISIBLE"

# Action
queue["action"] = "monitor"

# Highest-priority candidates receive refresh_review
queue.loc[
    queue["reason_code"].isin(["STALE_VISIBLE", "STALE_LOW_VISIBILITY"]),
    "action"
] = "refresh_review"

# Rank highest score first
queue = queue.sort_values(
    "score",
    ascending=False
).reset_index(drop=True)

queue["rank"] = queue.index + 1

# Create the required ranked queue
baseline_queue = queue[
    [
        "rank",
        "client_hash_id",
        "content_hash_id",
        "score",
        "reason_code",
        "action",
        "days_since_update",
        "impressions_30d",
        "clicks_30d",
        "avg_position"
    ]
].copy()

# Write the CSV required by the assignment
output_path = "work/outputs/baseline_action_score.csv"

os.makedirs("work/outputs", exist_ok=True)

baseline_queue.to_csv(output_path, index=False)

print(f"Saved: {output_path}")
print(f"Rows ranked: {len(baseline_queue):,}")

baseline_queue.head(10)
```

## 3. Top-20 review

*For each of the top 20: action, reason code, confidence note, and what would make it wrong.*

```python
# Review the top 20 baseline recommendations

top20 = baseline_queue.head(20).copy()

top20["confidence_note"] = top20.apply(
    lambda row: (
        "Higher confidence because the page is stale and has meaningful "
        "observed search visibility."
        if row["reason_code"] == "STALE_VISIBLE"
        else
        "Lower confidence because the page is stale but has limited "
        "observed search visibility."
        if row["reason_code"] == "STALE_LOW_VISIBILITY"
        else
        "Moderate confidence because the page has observed visibility "
        "but is not strongly stale."
        if row["reason_code"] == "RECENT_VISIBLE"
        else
        "Low confidence because neither strong staleness nor visibility "
        "is present."
    ),
    axis=1
)

top20["what_would_make_it_wrong"] = top20.apply(
    lambda row: (
        "The page may have been intentionally left unchanged, or the "
        "observed visibility may not justify a refresh."
        if row["reason_code"] == "STALE_VISIBLE"
        else
        "Low visibility may mean a refresh is unlikely to have enough "
        "search impact to justify the effort."
        if row["reason_code"] == "STALE_LOW_VISIBILITY"
        else
        "The page may be performing adequately despite being relatively "
        "recent, so a refresh may not be necessary."
        if row["reason_code"] == "RECENT_VISIBLE"
        else
        "The available signals may miss other reasons why the page "
        "should be reviewed."
    ),
    axis=1
)

top20_review = top20[
    [
        "rank",
        "content_hash_id",
        "action",
        "reason_code",
        "score",
        "days_since_update",
        "impressions_30d",
        "confidence_note",
        "what_would_make_it_wrong"
    ]
]

top20_review
```

## 4. Weak picks + leakage check

*Which picks look wrong and why? Confirm no product flags or future windows leaked in.*

## Weak picks + leakage check

Several top-ranked pages are weak picks because their high score is driven mainly by staleness while recent search visibility is very low or zero. These pages are still reasonable candidates for review under the baseline rule, but confidence is low because observed visibility is weak.

The strongest pick is the page with 264 days since update and 4,411 impressions because it combines substantial staleness with meaningful observed visibility.

Leakage check:
- The score uses `days_since_update` and `impressions_30d`.
- No future-window metrics or future labels are used.
- No product flags are used as score inputs.
- `action` is an output generated by the baseline rule, not an input to the score.

```python
# Leakage check

score_inputs = [
    "days_since_update",
    "impressions_30d"
]

output_columns = [
    "score",
    "reason_code",
    "action"
]

print("Score inputs:", score_inputs)

for col in score_inputs:
    print(f"{col}: present =", col in march_pages.columns)

print("\nOutput columns:", output_columns)

print("\nLeakage assessment:")
print("No future-window metrics or future labels were used.")
print("No product flags were used as score inputs.")
print("The action column is an output generated by the baseline rule, not a scoring input.")
```
